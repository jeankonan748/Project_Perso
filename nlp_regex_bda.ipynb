# Introduction aux expressions régulières



## Symboles de base: caractère alphanumérique

Tout d'abord, vous pouvez utiliser des expressions régulières comme une opération de recherche classique. Tous les **caractères alphanumériques** (lettre et chiffres), peuvent s'utiliser pour trouver leur semblable dans un corpus.


```python
import re
```

```python
text = \
"""
Voici un texte à essayer.
Saisissez n'importe quel mot de ce texte dans le paramètre de
la fonction `findall` prenant une expression régulière.
21/10/2025
"""
print(text)
```

Nous utilisons la fonction `findall` pour retrouver toutes les occurrences du motif ou chaîne de caractères recherchée.

```python
print(re.findall(r'te', text))
print(re.findall(r'texte', text))
```

```python
print(re.findall(r'1', text))
print(re.findall(r'2', text))
print(re.findall(r'20', text))
```

## Les disjonctions: usage de plage


Si nous voulions récupérer tous les mots composés de 5 caractères dans un texte. Comment s'y prendre?

> [abcdefghijklmnopqrstuvwxyz][abcdefghijklmnopqrstuvwxyz][abcdefghijklmnopqrstuvwxyz][abcdefghijklmnopqrstuvwxyz][abcdefghijklmnopqrstuvwxyz]

Pareil, pour un numéro de téléphone orange
> 07[06789][06789][0123456789][0123456789][0123456789][0123456789][0123456789][0123456789]

Très pesant !
Solution: les plages

> [a-z] = [abcdefghijklmnopqrstuvwxyz]

> [A-Z] =  Toute lettre majuscule

> [0-9] = [0123456789]

```python
text = \
"""
Ami, frère, collègue, jeune, âne,
0534236798
0769236790
0707236709
0504246780
07042467
"""
num_orange = r'07[06-9][06-9][0-9][0-9][0-9][0-9][0-9][0-9]'
mot_3 = r'[a-zèâ][a-zèâ][a-zèâ]'
```

```python
re.findall(num_orange, text)
```

```python
re.findall(mot_3, text)
```

## Répétition - Optionalité - Tout caractère:  `.` `?` `*` `+`

Tout caractère: `beg.n --> begin, began, begun, beg n, beg3n`

Optionalité: `oranges? --> orange, oranges`

Répétitions générales
- `*` signifie ≥ 0: `bo* --> b, bo, boo, booooo `

- `+` signifie ≥ 1: `bo+ --> bo, boo, booooo `

Répétitions personnalisées:
- Nombre spécifique `{n}` signifie exactement `n` fois
- Plage spécifique: `{min,max}` signifie de `min` fois à `max` fois
- Plage spécifique: `{min,}` signifie au moins `min` fois
- Plage spécifique: `{,max}` signifie au plus `max` fois


`n`, `min`, `max` sont tous entiers et `min ≤ max`


```python
num_orange = r'07[0-9]{6,8}'
mot_3 = r'[Aa-zèâ]{3}'
text = \
"""
Ami, frère, collègue, jeune, âne,
0534236798
0734236790
0704236709
0504246780
07042467
"""
```

```python
re.findall(num_orange, text)
```

```python
re.findall(mot_3, text)
```

## Négation de disjonctions: `^` étant 1er caractère entre les crochets`[]`

`^` signifie négation uniquement lorsqu'il est le 1er caractère dans ls crochets !!!

Les caractères spéciaux (`.` `?` `*` `+`) perdent leur sens (en regex) entre les crochets !!!

- `[^A-Z]` --> qui n'est pas une majuscule (comprends les chiffres !)
- `[^.]` --> qui n'est pas un point

```python
re.findall(r'[^?]', '123 adb2 ABB55B???')
```

## Alias ​​pratiques

| Motif (Pattern) | Expansion | Correspondances  |
|-----------------|------------|----------------------------|
| `\d` | `[0-9]` | Tout chiffre |
| `\D` | `[^0-9]` | Tout caractère non numérique |
| `\w` | `[a-zA-Z0-9_]` | Tout caractère alphanumérique ou _ |
| `\W` | `[^\w]` | Tout caractère non alphanumérique ou sans _ |
| `\s` | `[\r\t\n\f]` | Tout espace (espace, tabulation, retour à la ligne suivante) |
| `\S` | `[^\s]` | Tout caractère autre qu’un espace |

```python
re.findall(num_orange, text)
```

```python
re.findall(mot_3, text)
```

## Disjonction de motifs (patterns): `|`

`|` est peut être lu comme un `OU` logique mais il concerne la détection de deux motifs contrairement à `[]` qui concerne des caractères individuels

- `il|elle` --> `il, elle`
- `a|b|c` = `[abc]`
- `[Aa]ttiéké|[Pp]oisson` --> `Attiéké, attiéké, Poisson, poisson`




```python
num_orange = r'07\d{8}'
mot_3 = r'\w{3}'
text = \
"""
Ami, frère, collègue, jeune, âne,
Attiéké Poisson Fumé
attiéké poisson fumé
0534236798
0734236790
0704236709
0504246780
07042467
"""
```

```python
re.findall(f'{num_orange}|{mot_3}', text)
re.findall(r'07\d{8}|\w{3}', text)
```

```python
re.findall(rf'[Aa]ttiéké|[pP]oisson|[Ff]umé', text)
```

## Début et fin de ligne: `^`, `$`

- `^` désigne le début de ligne: `^[A-Z]` --> toute majuscule en début de ligne
- `^[^a-zA-Z]`: Tout caractère non alphabétique en début de ligne
- `.$`: Tout caractère en fin de ligne
- `\.$`: Tout point en fin de ligne

```python
text = \
"""
En Python, les expressions régulières (regex) sont gérées par le module `re`,
qui permet de trouver, d'extraire ou de remplacer des motifs de texte.
Elles sont utiles pour des tâches comme la validation de mots de passe
ou la recherche de motifs spécifiques dans des chaînes.
Mais aussi, elles servent à faire du scraping de texte.
Pour les utiliser, il faut d'abord importer le module `re`,
puis employer des fonctions comme `re.search()` pour trouver une occurrence,
`re.findall()` pour trouver toutes les occurrences,
ou `re.sub()` pour remplacer les motifs.
"""
text
```

```python
re.findall(r'^[A-Z]', text, re.MULTILINE)
```

## Caractères spéciaux hors des crochets

On précède les caractères spéciaux regex (`. ? * + []`) par un `\` lorsqu'on est hors des parenthèses:

- `\.$`: Tout point en fin de ligne
- `\?$`: point d'interrogation en fin de ligne.

## Sélection ou capture de groupes et substitutions

Vous pouvez regrouper des parties de la correspondance en les mettant entre parenthèses `()`. Ensuite, référencez les groupes dans l'expression régulière par `\1`, `\2`, etc. Par exemple, `(\w)\1` correspond à tout caractère alphanumérique répété deux fois (essayez ci-dessous).

Plus important encore, vous pouvez utiliser des groupes pour le remplacement, en les référençant par des nombres : `\1`, `\2`, etc. dans l'expression de remplacement.

```python
text = """John Smith
Jane Doe
Bob Thingum
John Appleseed
Bill Oddie
"""
print(text)
```

```python
regex = r'(\w+)\s(\w+)'
re.findall(regex, text)
```

```python
remplacement = r'\2 \1' # inversion prénom, nom --> nom, prénom
resultat = re.sub(regex, remplacement, text)
print(resultat)
```
